task 4

Write a function or pseudocode that takes a column of data (array of values) from a Flipkart order history and automatically detects whether it is likely Nominal, Ordinal, Interval, or Ratio scale. Briefly describe the logic used for each type.

```python
def detect_scale(column):
    values = [v for v in column if v is not None]

    if len(values) == 0:
        return "Unknown - empty column"

    # Text values
    if all(isinstance(v, str) for v in values):
        unique_vals = {v.lower() for v in values}

        ordinal_keywords = [
            {"low", "medium", "high"},
            {"small", "medium", "large"},
            {"placed", "shipped", "out for delivery", "delivered"},
            {"poor", "average", "good", "excellent"}
        ]

        for keyword_set in ordinal_keywords:
            if unique_vals.issubset(keyword_set) or unique_vals & keyword_set:
                return "Ordinal"

        return "Nominal"

    # Numeric values
    if all(isinstance(v, (int, float)) for v in values):
        has_true_zero = min(values) >= 0

        if is_cyclic_range(values) and 0 not in values:
            return "Interval"

        if has_true_zero:
            return "Ratio"

        if min(values) < 0:
            return "Interval"

    return "Unknown - mixed or unsupported type"


def is_cyclic_range(values):
    # crude check: bounded repeating range like hours (0-23) or degrees (0-360)
    return max(values) <= 360 and min(values) >= 0
```

```python
test_columns = {
    "Payment Method": ["UPI", "Card", "Cash", "UPI", "Net Banking"],
    "Order Status": ["Placed", "Delivered", "Shipped", "Out for Delivery"],
    "Number of Items Ordered": [1, 3, 0, 5, 2],
    "Amount Paid": [499.0, 1250.5, 0.0, 899.99, 2100.0],
    "Delivery Hour (24-hr)": [9, 14, 18, 23, 6]
}

for col, values in test_columns.items():
    print(f"{col:25s} -> {detect_scale(values)}")
```
